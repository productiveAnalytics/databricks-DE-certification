# Lecture - Introduction to Unit Tests for PySpark

## Overview

In this lecture, you’ll learn how to write and run unit tests for PySpark using pyspark.testing.utils and pytest, exploring their benefits, key functions, and practical examples to ensure code quality and reliability.

## Learning Objectives

By the end of this lecture, you will be able to: 
1. Learn to write unit tests for PySpark code to ensure functionality and catch errors early.
2. Apply the pytest framework in Databricks to execute unit tests and analyze the test results for errors.

<script>
(function () {
  const config = {
    mountId: "utd-mount",
    title: "3. Create the unit test",
    showAnnotationRail: true,
    hotspots: [
      { title: "1 · Define the test", startLine: 1, endLine: 1,
        description: "Create a unit test function to test the function. Name the unit test function accordingly (prefix with <code>test_</code>)." },
      { title: "2 · Sample DataFrame", startLine: 2, endLine: 4,
        description: "Create a sample DataFrame to test any use cases you can think of." },
      { title: "3 · Run the function", startLine: 6, endLine: 6,
        description: "Execute your function on the sample data and store the result." },
      { title: "4 · Expected result", startLine: 8, endLine: 11,
        description: "Create an expected result DataFrame using the sample data." },
      { title: "5 · Assert equality", startLine: 13, endLine: 13,
        description: "Check the two DataFrames. If they are not identical an error will be returned." },
    ],
    source: `def test_add_new_col():
    data = [(0,), (1,), (-1,),(None,)]
    columns = ["value"]
    df = spark.createDataFrame(data, columns)

    actual_df = add_new_col(df, "new_value", "value")

    expected_data = [(0, 'Normal'), (1, 'Unknown'),
                     (-1, 'Unknown'), (None, 'Unknown')]
    expected_df = spark.createDataFrame(expected_data,
                                        ["value", "new_value"])

    assertDataFrameEqual(actual_df, expected_df)`,
    showLineNumbers: true,
    displayStartLine: 1,
  };

  function escapeHtml(value) {
    return String(value).replace(/&/g, "&amp;").replace(/</g, "&lt;").replace(/>/g, "&gt;");
  }

  function renderSourceExplorer(config) {
    const mount = document.getElementById(config.mountId);
    const lines = config.source.replace(/\n$/, "").split("\n");
    const start = config.startLine || 1;
    const end = config.endLine || lines.length;
    const visible = lines.slice(start - 1, end);
    const hotspots = config.hotspots || [];
    const displayStartLine = config.displayStartLine || start;

    function hotspotFor(lineNo) {
      const index = hotspots.findIndex(h => lineNo >= h.startLine && lineNo <= h.endLine);
      return index >= 0 ? { hotspot: hotspots[index], index } : null;
    }

    const codeHtml = visible.map((line, idx) => {
      const actualLineNo = start + idx;
      const displayLineNo = displayStartLine + idx;
      const hotspot = hotspotFor(actualLineNo);
      const classes = hotspot ? `sx-line sx-hotspot sx-hotspot-${hotspot.index}` : "sx-line";
      const attrs = hotspot ? `data-hotspot-index="${hotspot.index}"` : "";
      return `<div class="${classes}" ${attrs}><div class="sx-line-no">${displayLineNo}</div><div class="sx-line-code">${escapeHtml(line)}</div></div>`;
    }).join("");

    const railHtml = config.showAnnotationRail && hotspots.length > 0
      ? `<div class="sx-rail">${hotspots.map((h, i) => `<button class="sx-chip" data-hotspot-index="${i}">${escapeHtml(h.title)}</button>`).join("")}</div>`
      : "";

    const overlayHtml = hotspots.length > 0
      ? `<div class="sx-overlay"><div class="sx-overlay-title"></div><div class="sx-overlay-body"></div></div>`
      : "";

    mount.innerHTML = `<div class="sx-wrap"><div class="sx-title">${escapeHtml(config.title)}</div>${railHtml}<div class="sx-code">${codeHtml}</div>${overlayHtml}</div>`;

    if (hotspots.length === 0) return;

    const overlay = mount.querySelector(".sx-overlay");
    const overlayTitle = mount.querySelector(".sx-overlay-title");
    const overlayBody = mount.querySelector(".sx-overlay-body");
    let activeHotspotIndex = null;

    function setActiveLines(idx) {
      mount.querySelectorAll(".sx-active-line").forEach(el => el.classList.remove("sx-active-line"));
      if (idx === null) return;
      mount.querySelectorAll(`.sx-hotspot-${idx}`).forEach(el => el.classList.add("sx-active-line"));
    }
    function setActiveChip(idx) {
      mount.querySelectorAll(".sx-chip").forEach(chip => chip.classList.toggle("active", idx !== null && Number(chip.dataset.hotspotIndex) === idx));
    }
    function setChipHover(idx, on) {
      mount.querySelectorAll(".sx-chip").forEach(chip => { const isMatch = Number(chip.dataset.hotspotIndex) === idx; chip.classList.toggle("sx-chip-hovered-from-line", on && isMatch); });
    }
    function setLineHover(idx, on) {
      mount.querySelectorAll(`.sx-hotspot-${idx}`).forEach(line => line.classList.toggle("sx-hovered-from-chip", on));
    }
    function showHotspot(idx, triggerEl = null) {
      const hotspot = hotspots[idx];
      if (!hotspot) return;
      if (overlay.style.display === "block" && activeHotspotIndex === idx) {
        overlay.style.display = "none"; activeHotspotIndex = null; setActiveChip(null); setActiveLines(null); return;
      }
      overlayTitle.textContent = hotspot.title;
      overlayBody.innerHTML = hotspot.description;
      overlay.style.display = "block";
      const wrap = mount.querySelector(".sx-wrap");
      const trigger = triggerEl || mount.querySelector(`.sx-hotspot-${idx}`) || mount.querySelector(`.sx-chip[data-hotspot-index="${idx}"]`);
      if (trigger) {
        const wrapRect = wrap.getBoundingClientRect();
        const triggerRect = trigger.getBoundingClientRect();
        const desiredTop = triggerRect.top - wrapRect.top;
        const maxTop = wrap.clientHeight - overlay.offsetHeight - 18;
        overlay.style.top = `${Math.max(18, Math.min(desiredTop, maxTop))}px`;
      }
      activeHotspotIndex = idx; setActiveChip(idx); setActiveLines(idx);
    }

    mount.querySelectorAll(".sx-hotspot").forEach(el => {
      const idx = Number(el.dataset.hotspotIndex);
      el.addEventListener("click", () => { showHotspot(idx, el); });
      el.addEventListener("mouseenter", () => { setLineHover(idx, true); setChipHover(idx, true); });
      el.addEventListener("mouseleave", () => { setLineHover(idx, false); setChipHover(idx, false); if (activeHotspotIndex !== null) setActiveLines(activeHotspotIndex); });
    });
    mount.querySelectorAll(".sx-chip").forEach(chip => {
      const idx = Number(chip.dataset.hotspotIndex);
      chip.addEventListener("click", () => { const associatedLine = mount.querySelector(`.sx-hotspot-${idx}`); showHotspot(idx, associatedLine || chip); });
      chip.addEventListener("mouseenter", () => { setLineHover(idx, true); });
      chip.addEventListener("mouseleave", () => { setLineHover(idx, false); if (activeHotspotIndex !== null) setActiveLines(activeHotspotIndex); });
    });
    mount.querySelector(".sx-wrap").addEventListener("click", (e) => {
      if (e.target.closest(".sx-hotspot")) return;
      if (e.target.closest(".sx-chip")) return;
      if (e.target.closest(".sx-overlay")) return;
      overlay.style.display = "none"; activeHotspotIndex = null; setActiveChip(null); setActiveLines(null);
    });
  }

  renderSourceExplorer(config);
})();
</script>

Next like talk about the pytest unit testing framework.<br>Pytest is a popular testing framework for Python that makes it easy to write simple and scalable test cases. It provides a variety of benefits for executing your unit tests.<br>First up, pytest has a very simple syntax. You don’t need to worry about complex setup. Just write test functions that start with test_, and pytest will automatically pick them up and run them.<br>Next, pytest leverages Python’s built-in assert statements (or you can use other assert statements like in pyspark). These are simple to use but provide detailed error messages when a test fails, making your function debugging much easier and faster by understanding what went wrong.<br>Pytest also automatically discovers and runs all your tests. There’s no need to manually configure which tests to run. Just name your test files and functions with the test_ prefix, and pytest will take care of the rest.<br>Lastly, pytest has a rich ecosystem of plugins to extend its functionality. Whether you need test coverage reports, parallel test execution, or integration with other tools, there&#x27;s a plugin for almost anything, allowing you to customize pytest to fit your needs.<br>This course provides a simple introduction to pytest. There are many testing frameworks available, select the one that best meets your organization&#x27;s needs.
