# Lecture - DevOps Fundamentals & the Role of CI/CD

## Overview

In this lecture, we explore what DevOps is and how it applies to Data Engineering, then dive into how CI/CD practices streamline the development, testing, and deployment of data pipelines. Together these ideas form the foundation for building reliable, automated Lakeflow Jobs.

## Learning Objectives

By the end of this lecture, you will be able to:
1. Explain how DevOps and DataOps principles integrate development, operations, and data workflows for smoother collaboration and faster delivery.
2. Describe the eight stages of the DevOps lifecycle and how they apply to Data Engineering and Machine Learning.
3. Outline the core concepts and benefits of CI/CD, focusing on automation, continuous integration, and continuous delivery/deployment of data pipelines.
4. Explain the testing pyramid and how continuous delivery differs from continuous deployment.

##### EXPAND FOR ADDITIONAL NOTES
<details>
<p>Let’s dive into the 8 key steps of the DevOps lifecycle. Click any stage of the loop to explore it.</p><p>First up is Planning — define project goals, gather requirements, and make sure the whole team is aligned on what needs to be delivered.</p><p>Next is Coding, where developers write the application’s source code, building the features and functionality we need.</p><p>After that comes Building, where we compile the code into executable files, making sure all dependencies are correctly integrated.</p><p>Then Testing — run automated tests to ensure the code works as expected, catching any bugs before we release. Testing is extremely important within DevOps.</p><p>Now it’s time for Release. We package the application, ensuring it’s production-ready, and prepare for a controlled rollout.</p><p>Once the release is ready, we Deploy — push the application to the production environment and make it available to users.</p><p>After deployment, we Operate the application: monitor performance, manage resources, and quickly address any issues.</p><p>Lastly, Monitoring — track the app’s performance, gather feedback, and continuously improve to keep things running smoothly.</p><p>The DevOps lifecycle is all about seamless collaboration between development and operations teams, continuously iterating as the project needs fixes, updates or features.</p>
</details>

##### EXPAND FOR ADDITIONAL NOTES
<details>
<p>Within CI/CD there are testing steps you should follow within what’s called the testing pyramid, which categorizes different tests: unit tests, integration tests and system tests.</p><p>The base of the pyramid are unit tests, which test individual functions or methods in isolation. They run quickly, frequently and automatically. Unit tests form the foundation because they are inexpensive and provide the broadest coverage — for example, testing if a pyspark method works as expected.</p><p>Next, integration tests test the interaction between different components or systems. These are slower and more costly, but provide greater assurance that components work together. Within Databricks these typically revolve around Notebooks, SDP and/or Lakeflow Jobs.</p><p>Lastly, system tests test the entire application in a real-world scenario. These are slow, costly, and often run in a production-like environment — for example, testing whether an end-to-end data pipeline works as expected within a Job.</p>
</details>

## M. Conclusion

<ul>
<li><b>DevOps</b> combines software engineering practices with IT operations to automate Lakeflow Jobs and streamline delivery across an eight-stage, continuous lifecycle.</li>
<li><b>DataOps</b> applies DevOps principles and culture to Data Engineering pipelines, turning business requirements into business value.</li>
<li><b>CI (Continuous Integration)</b> merges and automatically tests code changes often, following the testing pyramid of unit, integration, and system tests.</li>
<li><b>CD (Continuous Delivery/Deployment)</b> automates releases — delivery keeps a manual gate to production, while deployment pushes every passing change automatically.</li>
